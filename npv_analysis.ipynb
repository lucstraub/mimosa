Notebook to manually calculate NPV of mitigation costs over time

```python
import pandas as pd
import numpy as np

def npv(values, discount_rate):
    years = values.index.astype(float)
    t = years - years[0]
    discount_factor = np.exp(-discount_rate * t)
    return np.trapz(values * discount_factor, t)

outp = pd.read_csv(f"output/result_CE_budget500 GtCO2_industry0.4material0.5_CEcost200.0_run2025-05-23-15-34.csv")
# outp = pd.read_csv(f"output/CE_sensitivity_cost/result_CE_budget500 GtCO2_gammascale0.975_industry0.4material0.5_highCEscenarioTrue_CEcost1000.csv")
# outp = pd.read_csv(f"output/calibration_cost/result_nonCE_budget600 GtCO2_industry0.4_gammascale0.975.csv")
# outp = pd.read_csv(f"output/CE_sensitivity_abatement/result_CE_budget500 GtCO2_gammascale0.975_industry0.4material0.5_highCEscenarioFalse_CEabatement0.199.csv")
global_mitig_costs = outp.loc[outp["Variable"] == "mitigation_costs_regional", "2020":].sum(axis=0)
global_gdp_gross = outp.loc[outp["Variable"] == "GDP_gross", "2020":].sum(axis=0)
r = 0.03
npv_costs_abs = npv(global_mitig_costs, r)
npv_costs_rel = npv(global_mitig_costs, r) / npv(global_gdp_gross, r)
print(f"NPV of mitigation costs for this scenario: {npv_costs_abs:.1f} trillion*usd ({npv_costs_rel:.8%} of GDP)")
```

```python
def npv(values, discount_rate):
    years = values.index.astype(float)
    t = years - years[0]
    discount_factor = np.exp(-discount_rate * t)
    return np.trapz(values * discount_factor, t)

carbon_budgets = np.arange(510, 751, 10)
gamma_scaling = 0.975
results = []

for budget in carbon_budgets:
    outp = pd.read_csv(f"output/calibration_cost/result_nonCE_budget{budget} GtCO2_industry0.4_gammascale{gamma_scaling}.csv")
    global_mitig_costs = outp.loc[outp["Variable"] == "mitigation_costs_regional", "2020":].sum(
        axis=0
    )
    global_gdp_gross = outp.loc[outp["Variable"] == "GDP_gross", "2020":].sum(axis=0)
    r = 0.03
    npv_costs_abs = npv(global_mitig_costs, r)
    npv_costs_rel = npv(global_mitig_costs, r) / npv(global_gdp_gross, r)
    results.append({"budget": budget, "npv_costs_abs": npv_costs_abs, "npv_costs_rel": npv_costs_rel})
    print(f"NPV of mitigation costs for {budget} GtCO2 non-CE scenario: {npv_costs_abs:.1f} trillion*usd ({npv_costs_rel:.8%} of GDP)")

results = pd.DataFrame(results)
# results.to_csv(f"output/npv_costs_nonCE_industry0.35_gammascale{gamma_scaling}.csv", index=False)
```
